```python
import config as cfg
from environment import CombatEnv
from A2C_Agents import A2CAgent
```

```python
env = CombatEnv()
agent = A2CAgent(obs_dim=7)
```

```python
EPISODES = 5000
wins, losses, ties = 0, 0, 0
```

```python
for ep in range(EPISODES):
    state = env.reset()
    done = False

    while not done:
        agent_action = agent.select_action(state)
        player_action = env.player.choose_action_probablistic(env.agent.action_streak)

        state, reward, done, info = env.step(agent_action, player_action)
        agent.record_reward(reward)

    agent.learn()

    if env.player.hp <= 0 and env.agent.hp > 0: wins += 1
    elif env.agent.hp <= 0 and env.player.hp > 0: losses += 1
    else: ties += 1

    if (ep + 1) % 500 == 0:
        print(f"Ep {ep+1}: WinRate {wins/(ep+1):.1%} | LossRate {losses/(ep+1):.1%} | Ties {ties}")

agent.save()
```
